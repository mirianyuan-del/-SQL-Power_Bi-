```python
import pandas as pd
from sqlalchemy import create_engine

# 1. 读原始数据
df = pd.read_csv('Shopping Trends And Customer Behaviour Dataset.csv')

# 2. 清洗（幂等写法，重复执行也不报错）
df = df.loc[:, ~df.columns.str.startswith('Unnamed')]      # 去掉索引残留列
df.columns = df.columns.str.lower().str.replace(' ', '_')  # 列名转小写蛇形
if 'promo_code_used' in df.columns:
    df = df.drop('promo_code_used', axis=1)
labels = ['Young Adult', 'Adult', 'Middle Age', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)

# 3. 连接并重建表（关键：清洗和写入在同一个单元格，顺序不会乱）
engine = create_engine("mysql+pymysql://root:01727abc1588@localhost:3306/shop_db?charset=utf8mb4")
df.to_sql('shopping_trends', engine, if_exists='replace', index=False)

print(df.columns.tolist())
print("写入完成，共", len(df), "行")
```

```python
pd.read_sql("SELECT * FROM shopping_trends LIMIT 1", engine).columns.tolist()
```
